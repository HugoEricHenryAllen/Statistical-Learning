# Logistic Regression

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('insurance_data.csv')

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111)
ax.set(xlabel='age', ylabel='bought_insurance')
plt.scatter(df['age'], df['bought_insurance'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(df[['age']], df[['bought_insurance']], test_size=0.1)
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, Y_train.values.ravel())
```

```python
import numpy as np

fig = plt.figure(figsize=(8,8))

X = np.linspace(min(df['age']),max(df['age']), max(df['age'])-min(df['age'])+1)
X = pd.DataFrame(X, columns=['age'])


ax = fig.add_subplot(111)
ax.set(xlabel='age', ylabel='bought_insurance', title='Logistic Regression for age-insurance data, blue line is the model decision, red line is model probability')
ax.plot(X, model.predict(X), color='blue')
ax.plot(X, model.predict_proba(X)[:,1], color='red')
ax.scatter(df['age'], df['bought_insurance'])
```
