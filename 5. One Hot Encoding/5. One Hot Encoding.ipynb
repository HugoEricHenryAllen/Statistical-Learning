# Dummy Variable One Hot Encoding
using the following sklearn methods:\
from sklearn.compose import ColumnTransformer\
from sklearn.preprocessing import OneHotEncoder\
from sklearn.preprocessing import LabelEncoder

```python
df = pd.read_csv('carprices.csv')
df
```

We create dummy variables for each category in the categorical feature vector of 'Car Model'

```python
dummies = pd.get_dummies(df['Car Model'],dtype=int)
merged = pd.concat([df, dummies], axis=1)
merged
```

Now, to clean up our data frame, we want to remove the categorical feature vector 'Car Model'. And also remove one of the dummy variable columns to avoid the dummy variable trap. See https://www.kaggle.com/discussions/general/134922

```python
final = merged.drop(['Car Model', 'Mercedez Benz C class'], axis=1)
final
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

Initialising Input and Output data frames

```python
X = final.drop(['Sell Price($)'], axis=1)
Y = final['Sell Price($)']
X
```

```python
model.fit(X,Y)
```

```python
car = pd.DataFrame(
    [[69000, 4, 1, 0]],
    columns=X.columns
)
model.predict(car)
```

Alternatively, we can use Sklearn to perform One Hot encoding of our data frame

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
dfle = df
dfle['Car Model'] = le.fit_transform(dfle['Car Model'])
dfle
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

X = df[['Car Model', 'Mileage', 'Age(yrs)']]
Y = df['Sell Price($)']

ct = ColumnTransformer(
    [('car_model', OneHotEncoder(sparse_output=False),
      ['Car Model'])],
    remainder='passthrough'
)

X_encoded = ct.fit_transform(X)
X_encoded
```

```python
X_encoded = X_encoded[:,1:] #drop first dummy variable
```

```python
model.fit(X_encoded, Y)
```

```python
model.predict([[1,0, 69000, 4]])
```
