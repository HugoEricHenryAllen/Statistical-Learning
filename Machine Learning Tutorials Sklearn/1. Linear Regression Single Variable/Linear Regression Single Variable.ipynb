# Linear Regression Single Variable

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("canada_per_capita_income.csv")
df[['year']].values.ravel()
```

```python
plt.scatter(df[['year']], df[["per capita income (US$)"]])
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['year']].values, df[["per capita income (US$)"]].values)
```

```python
reg.predict([[2020]])
```

```python
print(reg.coef_, reg.intercept_)
```

```python
years = np.array([[2018], [2020], [2022], [2025]])
```

```python
reg.predict(df[['year']].values).ravel()
```

```python
results = pd.DataFrame({'year' : years.ravel(), "per capita income (US$)" : reg.predict(years).ravel()})
```

```python
plt.xlabel('year', fontsize=15)
plt.ylabel('per capita income (US$)', fontsize=15)

plt.scatter(df[['year']], df[["per capita income (US$)"]])
plt.plot(df[['year']].values.ravel(), reg.predict(df[['year']].values).ravel(), color='red')
plt.plot(years.ravel(), reg.predict(years).ravel(), color='green')
```

```python
results.to_csv("predicted_income.csv")
```
